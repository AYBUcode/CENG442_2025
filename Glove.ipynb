```python
from gensim.scripts.glove2word2vec import glove2word2vec
from gensim.models import KeyedVectors
```

```python
glove_input = 'glove.6B.100d.txt'
word2vec_output = 'glove.6B.100d.word2vec'

# Load the GloVe vectors directly using load_word2vec_format
model = KeyedVectors.load_word2vec_format(glove_input, binary=False, no_header=True)
```

```python
model['istanbul']
```

```python
model.most_similar('elephant')
```

```python
model.most_similar(positive=['woman', 'son'], negative=['man'], topn=1)
```

```python
model.most_similar(positive=['bangkok', 'germany'], negative=['berlin'], topn=1)
```
